# Small Dataset

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
while not (PROJECT_ROOT / "pyproject.toml").is_file():
    if PROJECT_ROOT.parent == PROJECT_ROOT:
        raise RuntimeError("Raiz do projeto não encontrada.")
    PROJECT_ROOT = PROJECT_ROOT.parent

SRC_DIR = PROJECT_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

PROJECT_ROOT
```

```python
from classical.runner import run_experiment

result = run_experiment(
    dataset_config_path=PROJECT_ROOT / "src/config/dataset/small.yaml",
    model_config_path=PROJECT_ROOT / "src/config/model/lenet.yaml",
    project_root=PROJECT_ROOT,
    overrides={
        "training": {"epochs": 3},
        "dataset": {"batch_size": 8, "num_workers": 0},
        "output": {"run_name": "notebook_small_test"},
    },
)
```

```python
result["output_dir"], result["metrics"]
```
